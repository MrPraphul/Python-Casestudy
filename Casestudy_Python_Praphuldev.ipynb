```python
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]

total_employees = len(employees)
print(f"\nTotal Employees: {total_employees}")

salaries = [employee["salary"] for employee in employees]

average_salary = sum(salaries) / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print(f"\nAvarage Salary: {average_salary:.1f}")
print(f"Highest Salary: {highest_salary}")
print(f"Lowest Salary {lowest_salary}")

print("\nEmployees with Salary > 65000 and Rating >= 4.5")
for employee in employees:
    if employee["salary"] > 65000 and employee["rating"] >= 4.5:
      print(employee["name"])


employee_id_name_dict = {employee["id"]: employee["name"] for employee in employees}
print("\nEmployee ID-Name Dictionary")
print(employee_id_name_dict)

employee_names = [employee["name"] for employee in employees]
employee_names.sort()
employee_names.reverse()

print("\nFirst Three Employee Names (Reverse Alphabetical)")
for name in employee_names[:3]:
    print(name)

employee_ids = tuple(employee["id"] for employee in employees)

print("\nEmployee ID Tuple")
print(employee_ids)
print(f"Maximum Employee ID: {max(employee_ids)}")
print(f"Minimum Employee ID: {min(employee_ids)}")

unique_departments = {employee["department"] for employee in employees}

print("\nDepartments")
print(unique_departments)

employee_id_to_search = int(input("Enter Employee ID to search: "))

found_employee = None
for employee in employees:
  if employee["id"] == employee_id_to_search:
        found_employee = employee
        break

if found_employee:
    print("\nEmployee Found")
    print(f"ID: {found_employee['id']}")
    print(f"Name: {found_employee['name']}")
    print(f"Department: {found_employee['department']}")
    print(f"Salary: {found_employee['salary']}")
    print(f"Rating: {found_employee['rating']}")
else:
    print("Employee Not Found")
```
